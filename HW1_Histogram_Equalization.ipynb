```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('histogram_equalization_img.png', cv2.IMREAD_GRAYSCALE)

# HE, CLAHE
he = cv2.equalizeHist(img)
clahe = cv2.createCLAHE(clipLimit=4.0, tileGridSize=(8, 8)).apply(img)

titles = ['Original', 'HE', 'CLAHE']
images = [img, he, clahe]

plt.figure(figsize=(7, 7))

for i in range(3):
    plt.subplot(3, 2, 2*i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

    # histogram
    plt.subplot(3, 2, 2*i+2)
    hist = cv2.calcHist([images[i]], [0], None, [256], [0, 256])
    plt.plot(hist, color='black')
    plt.xlim([0, 256])
    plt.title(f'{titles[i]} Histogram')

plt.tight_layout()
plt.show()
```
