# Arithmetic operations

## 1. Addition

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('arithmetic_img1.png')
img2 = cv2.imread('arithmetic_img2.png')

if img1.shape != img2.shape:
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# addition
added = cv2.add(img1, img2)

plt.subplot(1, 3, 1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Added")
plt.imshow(cv2.cvtColor(added, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 2. Subtraction

```python
img3 = cv2.imread('arithmetic_img3.png')
img4 = cv2.imread('arithmetic_img4.png')

if img3.shape != img4.shape:
    img4 = cv2.resize(img4, (img3.shape[1], img3.shape[0]))

# subtraction
subtracted = cv2.subtract(img3, img4)

plt.subplot(1, 3, 1)
plt.title("Image 3")
plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 4")
plt.imshow(cv2.cvtColor(img4, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Subtracted")
plt.imshow(cv2.cvtColor(subtracted, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 3. Division

```python
img5 = cv2.imread('arithmetic_img5.png')
img6 = cv2.imread('arithmetic_img6.png')

if img5.shape != img6.shape:
    img6 = cv2.resize(img6, (img5.shape[1], img5.shape[0]))

# division
# division 결과가 0 ~ 1 사이의 값이 대부분이기 때문에 scale 하지 않을 경우 거의 검정색처럼 보임 -> 정규화 효과
divided = cv2.divide(img5, img6, scale=255.0)

plt.subplot(1, 3, 1)
plt.title("Image 5")
plt.imshow(cv2.cvtColor(img5, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 6")
plt.imshow(cv2.cvtColor(img6, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Divided")
plt.imshow(cv2.cvtColor(divided, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 4. Multiplication

```python
img7 = cv2.imread('arithmetic_img7.png')

# multiplication
alpha = 1.3
multiplied = cv2.multiply(img7, alpha)

plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(cv2.cvtColor(img7, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title(f"Multiplied (x{alpha})")
plt.imshow(cv2.cvtColor(multiplied, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```
