```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('logical_img1.png', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('logical_img2.png', cv2.IMREAD_GRAYSCALE)

if img1.shape != img2.shape:
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# binarization (threshold: 127 이하 → 0, 초과 → 255)
_, bin1 = cv2.threshold(img1, 127, 255, cv2.THRESH_BINARY)
_, bin2 = cv2.threshold(img2, 127, 255, cv2.THRESH_BINARY)

# logical operations
and_result = cv2.bitwise_and(bin1, bin2)
or_result  = cv2.bitwise_or(bin1, bin2)
xor_result = cv2.bitwise_xor(bin1, bin2)
not_result = cv2.bitwise_not(bin1)

titles = [
'Image 1', 'Binary 1', 'Image 2', 'Binary 2',
    'AND', 'OR', 'XOR', 'NOT'
]
images = [
    img1, bin1, img2, bin2,
    and_result, or_result, xor_result, not_result
]

plt.figure(figsize=(7, 5))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
