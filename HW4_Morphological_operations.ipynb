# Morphological operations

## 1. Dilation and Erosion

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('flower.png', cv.IMREAD_GRAYSCALE)

kernel = np.ones((5,5), np.uint8)

# dialation and erosion
dilation = cv.dilate(img, kernel, iterations=1)
erosion = cv.erode(img, kernel, iterations=1)

plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Dilation")
plt.imshow(dilation, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Erosion")
plt.imshow(erosion, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 2. Opening and Closing


```python
img_open = cv.imread('j_opening.png', cv.IMREAD_GRAYSCALE)
img_close = cv.imread('j_closing.png', cv.IMREAD_GRAYSCALE)

kernel = np.ones((5, 5), np.uint8)

# opening and closing
opened = cv.morphologyEx(img_open, cv.MORPH_OPEN, kernel) # 배경보다 작은 작은 점/노이즈 제거
closed = cv.morphologyEx(img_close, cv.MORPH_CLOSE, kernel) # 객체 내부의 작은 구멍 채우기, 경계 다듬기

plt.figure(figsize=(14, 4))

plt.subplot(1, 4, 1)
plt.title("Opening Original")
plt.imshow(img_open, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title("Opening Result")
plt.imshow(opened, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title("Closing Original")
plt.imshow(img_close, cmap='gray')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title("Closing Result")
plt.imshow(closed, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
